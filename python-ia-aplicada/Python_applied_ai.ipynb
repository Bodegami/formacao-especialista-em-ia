## Lendo e escrevendo em arquivos python

```python
def resumidor_de_emails(lista_de_emails):
  lista_de_resumos = []

  for i, email in enumerate(lista_de_emails):
    resposta = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=f""" Vou te mandar o corpo de um e-mail. Quero que você o resuma em apenas 1 linha,
          passando o intuito daquele e-mail.
          Segue o e-mail: {email}
          """
        )

    resumo = f"E-mail {i + 1}: {resposta.text}"
    print(resumo)
    lista_de_resumos.append(resumo)
    print("-" * 50)

  return lista_de_resumos
```

```python
lista_resumos = resumidor_de_emails(emails)
```

```python
with open("lista-de-resumos.txt", "w", encoding="utf-8") as arquivo:
  for resumo in lista_resumos:
    arquivo.write(resumo + "\n")
```

```python
from groq import Groq

client = Groq(api_key=os.environ.get("GROQ_KEY"))
```

```python
def tira_duvidas(lista_de_duvidas):
    lista_de_respostas = []

    for i, pergunta in enumerate(lista_de_duvidas):
        pergunta_formatada = f"""Responda a minha pergunta de forma resumida e no máximo de 1 linha.
        {pergunta}
        """

        completion = client.chat.completions.create(
            model="llama-3.1-8b-instant",
            messages=[
                {
                    "role": "user",
                    "content": pergunta_formatada
                }
            ],
            temperature=1,
            max_completion_tokens=1024,
            top_p=1,
            stream=True,
            stop=None
        )

        resposta_completa = ""
        print(f"Resposta {i + 1}: ", end="")

        for chunk in completion:
            texto_chunk = chunk.choices[0].delta.content or ""
            print(texto_chunk, end="")
            resposta_completa += texto_chunk

        print("\n")

        lista_de_respostas.append(resposta_completa)

    return lista_de_respostas
```

```python
# Etapa 1
lista_de_perguntas = [
    "Qual é a capital da Austrália?",
    "Quem pintou a 'Mona Lisa'?",
    "Quantos planetas existem no Sistema Solar?",
    "Em que ano terminou a Segunda Guerra Mundial?",
    "Qual é o rio mais longo do mundo?",
    "Quem escreveu 'Dom Casmurro'?",
    "Qual elemento químico tem o símbolo 'O'?",
    "Qual é o animal terrestre mais rápido do mundo?",
    "Quantos ossos tem um corpo humano adulto?",
    "Em qual continente fica o Deserto do Saara?"
]

with open("lista-de-perguntas-professor.txt", "w", encoding="utf-8") as arquivo:
  for pergunta in lista_de_perguntas:
    arquivo.write(pergunta + "\n")
```

```python
# Etapa 2
lista_desafio = []

with open("lista-de-perguntas-professor.txt", "r", encoding="utf-8") as arquivo:
  for linha in arquivo:
    lista_desafio.append(linha.strip())

print(lista_desafio)
```

```python
# Etapa 3
def busca_respostas(lista_desafio):
    lista_de_dicionarios = []

    for pergunta in lista_desafio:
        pergunta_formatada = f"""Gere uma resposta muito sucinta e em 1 linha sobre a minha pergunta.
        {pergunta}
        """

        completion = client.chat.completions.create(
            model="llama-3.1-8b-instant",
            messages=[
                {
                    "role": "user",
                    "content": pergunta_formatada
                }
            ],
            temperature=1,
            max_completion_tokens=1024,
            top_p=1,
            stream=True,
            stop=None
        )

        resposta_completa = ""
        for chunk in completion:
            texto_chunk = chunk.choices[0].delta.content or ""
            resposta_completa += texto_chunk

        resposta = {"pergunta": pergunta, "resposta": resposta_completa}
        lista_de_dicionarios.append(resposta)

    return lista_de_dicionarios
```

```python
lista_respostas_professor = busca_respostas(lista_desafio=lista_desafio)
```

```python
print(lista_respostas_professor)
```

```python
# Etapa 4
with open("lista-de-respostas-professor.csv", mode="w", encoding="utf-8") as arquivo:
  arquivo.write("pergunta,resposta\n")

  for pergunta_dict in lista_respostas_professor:
    arquivo.write(f"{pergunta_dict['pergunta']},{pergunta_dict['resposta']}\n")
```

```python
# Etapa 4 usando pandas
df_perguntas_respostas = pd.DataFrame(lista_respostas_professor)

df_perguntas_respostas.to_csv("lista-de-respostas-professor-pandas.csv", index=False, encoding="utf-8")
```
