Desafio: Crie um código que percorra a lista de médias e some 1 em cada valor, exceto quando o valor passar de 10.0, que aí ele deve parar em 10.0

```python
for media in lista_de_medias:
  media += 1
  if media > 10.0:
    media = 10.0
  print(media)
```

```python
n = 0
lista_de_medias = [8.9, 7.5, 4.2, 1.4, 9.5]

while n < len(lista_de_medias):
  lista_de_medias[n] += 1.0
  if lista_de_medias[n] > 10.0:
    lista_de_medias[n] = 10.0

  n += 1

print(lista_de_medias)

```

*Dicionario*</br> chave: valor

```python
lista_de_nomes = ["João CLeber", "Maria Silva", "Pedro Afonso", "Daniel Lopes", "Debora Carvalho"]
lista_de_medias = [8.9, 7.5, 4.2, 1.4, 9.5]
```

```python
dict_nomes_medias = {}
for nome, media in zip(lista_de_nomes, lista_de_medias):
  dict_nomes_medias[nome] = media
```

```python
print(dict_nomes_medias)
```

```python
dict_nomes_medias["João CLeber"]
```

```python
dict_nomes_medias.get("Pedro Afonso")
```

```python
dict_nomes_medias.pop("Debora Carvalho")
```
