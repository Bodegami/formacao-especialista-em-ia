```python
if idade >= 18:
    print(f"O {nome} pode entrar na festa porque ele é maior de idade.")
    print(f"Ele tem {idade} anos de idade.")
else:
    print(f"O {nome} NÃO pode entrar na festa porque ele é menor de idade.")
    print(f"Ele tem {idade} anos de idade.")
```

```python
media = float(input("Digite a sua media: "))

if media >= 7.0:
  print("Aprovado")
elif media >= 5.0:
  print("Recuperação")
else:
  print("Reprovado")
```

```python
media = float(input("Digite a sua media: "))

if media < 5.0:
  print("Reprovado")
elif media >= 5.0 and media < 7.0:
  print("Recuperação")
else:
  print("Aprovado")
```

# Rodando a sua IA no Google Colab

```python
import os
from google.colab import userdata

os.environ['GOOGLE_API_KEY'] = userdata.get('GEMINI_KEY')
```

```python
from google import genai

client = genai.Client()
```

```python
resposta = client.models.generate_content(model="gemini-3.5-flash", contents="O que é a Inteligência Artificial?")
```

```python
print(resposta.text)
```
