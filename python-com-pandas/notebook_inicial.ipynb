## Filtros

### **1. Apartamentos que possuem `1 quarto` e `aluguel menor que 1200`**



```python
df['Quartos'] == 1
```

```python
selecao1 = df['Quartos'] == 1
df[selecao1] # Apresenta os apartamentos que possuem 1 quarto
```

```python
selecao2 = df['Valor'] < 1200
df[selecao2] # Apresenta os apartamentos que possuem aluguel menor que 1200
```

```python
selecao_final = (selecao1) & (selecao2)
df[selecao_final] # Apresenta os apartamentos que possuem 1 quarto e aluguel menor que 1200
```

```python
df_1 = df[selecao_final] # Cria um novo dataframe com os apartamentos que possuem 1 quarto e aluguel menor que 1200
df_1.head()
```

### **2. `Apartamentos` que possuem pelo menos `2 quartos`, `aluguel menor que 3000` e `area maior que 70`**

```python
selecao = (df['Quartos'] >= 2) & (df['Valor'] < 3000) & (df['Area'] > 70)
df[selecao] # Apresenta os apartamentos que possuem pelo menos 2 quartos, aluguel menor que 3000 e area maior que 70
```

```python
df_2 = df[selecao] # Cria um novo dataframe com os apartamentos que possuem pelo menos 2 quartos, aluguel menor que 3000 e area maior que 70
df_2.head()
```

# Manipulando os dados

## Criando colunas numéricas

```python
url = 'https://raw.githubusercontent.com/alura-cursos/pandas-conhecendo-a-biblioteca/main/base-de-dados/aluguel.csv'
dados = pd.read_csv(url, sep=';')
dados.head()
```

```python
dados['Valor_por_mes'] = dados['Valor'] + dados['Condominio'] # Insere uma nova coluna que representa a soma do Valor e do Condominio por mes
dados.head()
```

```python
dados['Valor_por_ano'] = dados['Valor_por_mes'] * 12 + dados['IPTU'] # Insere uma nova coluna que representa o Valor anual e o Iptu
dados.head()
```

## Criando colunas categóricas

```python
dados['Descricao'] = dados['Tipo'] + ' em ' + dados['Bairro'] # Insere uma nova coluna que representa a descrição do imóvel
dados.head()
```

```python
dados['Descricao'] = dados['Tipo'] + ' em ' + dados['Bairro'] + ' com ' + \
                      dados['Quartos'] + ' quarto(s)' + \
                      ' e ' + dados['Vagas'] + ' vaga(s) de garagem.'
```

```python
dados['Descricao'] = dados['Tipo'] + ' em ' + dados['Bairro'] + ' com ' + \
                      dados['Quartos'].astype(str) + ' quarto(s)' + \
                      ' e ' + dados['Vagas'].astype(str) + ' vaga(s) de garagem.'
dados.head()
```
